```python
UPDATED MATLAB SOLVER FOR THE HYDROGEL MANUSCRIPT
=================================================

Purpose
-------
This folder implements the revised main.tex model:
  1. Stokes flow in the media channel.
  2. Biot pore-pressure evolution in the GelMA region.
  3. Quasi-static small-strain skeleton displacement.
  4. One-mode Standard Linear Solid stress update.
  5. Darcy seepage/pore velocity.
  6. Solute advection-diffusion across gel/post/media regions.
  7. UV radiant-exposure -> porosity -> permeability -> Deff update.
  8. Updated non-dimensional groups.

Run
---
Open MATLAB in this folder and execute:

    run_all

The scripts are written for MATLAB R2021a or newer and do not require a special toolbox.

Main outputs
------------
fig3_concentration.png  : baseline normalized concentration field
fig4_UV_sweep.png       : UV radiant-exposure sweep (keeps old filename for LaTeX)
fig_pressure.png         : final GelMA pore pressure
fig_stress.png           : final equivalent SLS effective stress
baseline_results.mat     : all fields and diagnostics
uv_sweep_results.mat     : dose-sweep results

Optional checks
---------------
    verify_convergence
    run_parameter_sensitivity

CRITICAL PARAMETER STATUS
-------------------------
Directly anchored to supplied experimental/source material:
  Geometry L, Wg, Wm, d, H
  UV wavelength = 365 nm
  Medium-bloom 5 wt% E = 0.18 kPa (baseline formulation)
  E=3G conversion used in the JMBBM paper
  Permeability order 1e-17 to 1e-16 m^2
  Poroelastic diffusivity order 1e-12 to 1e-11 m^2/s
  Mesh size ~4 nm for medium/high bloom formulations
  Saturated-gel alpha=1 is used as the stated baseline assumption

Still provisional/assumed because the supplied documents do not give calibrated numbers:
  actual culture-medium viscosity at 37 C
  actual pump mean velocity / flow rate
  G0, Ginf and relaxation time from raw indentation fits
  drained bulk modulus Kdr
  Biot modulus M
  volumetric phi0 from raw wet/dry mass data
  UV irradiance and fitted beta_H
  solute-specific Ds0, Ds_eff, hydrodynamic radius and Amsden coefficient

IMPORTANT DIFFUSIVITY DISTINCTION
---------------------------------
Dp in the JMBBM paper is poroelastic solvent/water redistribution diffusivity.
It is NOT the molecular diffusivity of a fluorescent tracer/protein/drug.
The code keeps Dp separate from Ds0 and Ds_eff.

MODEL CLOSURES THAT SHOULD BE REVISITED WHEN NEW DATA ARRIVE
------------------------------------------------------------
1. Solid boundary conditions are currently an anchored rectangular GelMA block.
   The manuscript does not yet specify experimentally constrained solid tractions.
2. The micropost row is homogenized as a transition strip, not resolved post by post.
3. UV porosity law beta_H is a placeholder until porosity-vs-dose data are fitted.
4. The one-mode SLS is a reduction of a two-mode generalized Maxwell fit.
5. Kdr and M are assumptions and should be subjected to sensitivity analysis.

HOW TO INSERT WET/DRY MASS DATA
-------------------------------
Use:

    phi = porosity_from_swelling(ms,md,rho_liquid,rho_polymer)

where ms and md use the same mass units. Then put the resulting mean in p.phi0.

HOW TO INSERT ACTUAL FLOW RATE
------------------------------
If the pump gives volumetric rate Q [m^3/s], an approximate mean media-channel velocity is

    U0 = Q/(Wm*H)

for one media channel. Set p.U0 to that value.

HOW TO INSERT UV DOSE
---------------------
For a single exposure:

    H_UV [mJ/cm^2] = irradiance [mW/cm^2] * time [s]

Set p.uv_irradiance_mW_cm2 and p.uv_exposure_s accordingly.
```
